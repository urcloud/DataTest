#### 데이터 불러오기 (Load the data)

이번 노트북에서는 다음 6개 국가의 FIFA 랭킹 데이터를 사용합니다.
* 아르헨티나(ARG)
* 브라질(BRA)
* 스페인(ESP)
* 프랑스(FRA)
* 독일(GER)
* 이탈리아(ITA)

데이터는 CSV 파일(Comma-Separated Values)에 저장되어 있습니다.

CSV 파일을 Excel로 열어보면:
* 각 행(row)은 날짜
* 각 열(column)은 국가
를 의미합니다.



데이터를 불러오는 과정은 2단계입니다.
1. 데이터 파일 위치(filepath) 지정
2. 해당 파일을 노트북으로 읽어오기

```python
# Path of the file to read
fifa_filepath = "../input/fifa.csv"

# Read the file into a variable fifa_data
fifa_data = pd.read_csv(fifa_filepath, index_col="Date", parse_dates=True)
```



#### 코드 설명

2. 실행 코드(Executable code)

실제로 컴퓨터가 실행하는 코드입니다.

* 첫 번째 실행 코드

```python
fifa_filepath = "../input/fifa.csv"
```
데이터 파일 위치를 fifa_filepath 변수에 저장합니다.

* 두 번째 실행 코드

```python
fifa_data = pd.read_csv(fifa_filepath, index_col="Date", parse_dates=True)
```
CSV 파일 내용을 fifa_data 변수에 저장합니다.

여기서:
* fifa_filepath
→ 읽어올 파일 경로
* index_col="Date"
→ "Date" 열을 행의 기준(index)으로 사용
* parse_dates=True
→ 날짜 데이터를 날짜 형식으로 인식

이 코드들은 출력 결과가 나타나지 않습니다.

#### 데이터 시각화하기 (Plot the data)

이 강의에서는 다양한 그래프를 배우게 됩니다.
놀랍게도 대부분의 그래프는 단 한 줄의 코드로 만들 수 있습니다!
아래는 선 그래프(Line Chart)를 만드는 코드입니다.

```python
# Set the width and height of the figure
plt.figure(figsize=(16,6))

# Line chart showing how FIFA rankings evolved over time 
sns.lineplot(data=fifa_data)
```
