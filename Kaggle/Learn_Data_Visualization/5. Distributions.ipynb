#### Histogram (히스토그램)

히스토그램은:
데이터가 어떤 값에 얼마나 많이 분포하는지
보여주는 그래프입니다.

#### 예제 목표

꽃잎 길이(Petal Length)가:
어떤 분포를 가지는지 확인

#### 코드

```python
sns.histplot(
    iris_data['Petal Length (cm)']
)
```



#### 설명

* sns.histplot()
히스토그램 생성 함수

* 입력 데이터
```python
iris_data['Petal Length (cm)']
```
꽃잎 길이 컬럼

#### 그래프 해석

막대가 높을수록:
해당 길이를 가진 꽃이 많음
을 의미합니다.

#### KDE Plot (밀도 그래프)

다음은 KDE Plot입니다.

KDE는:
부드럽게 표현한 히스토그램
이라고 생각하면 됩니다.

#### 코드

```python
sns.kdeplot(
    data=iris_data['Petal Length (cm)'],
    shade=True
)
```



#### 설명

* sns.kdeplot()
밀도 곡선 생성

* shade=True
곡선 아래 영역 색칠
(최신 seaborn에서는 fill=True 사용 권장)

#### 경고 메시지 의미

```python
shade is deprecated in favor of fill
```

의미:
* 앞으로는 shade=True
* 대신 fill=True
사용 권장

#### 최신 코드 형태

```python
sns.kdeplot(
    data=iris_data['Petal Length (cm)'],
    fill=True
)
```

#### 그래프 해석

곡선이 높을수록:
해당 길이의 꽃이 많이 존재

#### 2D KDE Plot

이번에는:
두 변수의 분포를 동시에 시각화
합니다.

#### 목표
* x축 → 꽃잎 길이
* y축 → 꽃받침 너비
관계 확인

#### 코드

```python
sns.jointplot(
    x=iris_data['Petal Length (cm)'],
    y=iris_data['Sepal Width (cm)'],
    kind="kde"
)
```



#### 설명

* sns.jointplot()
두 변수 관계 시각화

* kind="kde"
2D KDE plot 생성

#### 그래프 해석

색이 진할수록:
해당 조합이 자주 나타남

#### 추가로 보이는 그래프들

중앙 2D KDE 외에도:

#### 위쪽 그래프

x축 변수의 KDE

즉:
```python
Petal Length
```
분포

#### 오른쪽 그래프

y축 변수의 KDE

즉:
```python
Sepal Width
```
분포

#### Color-coded plots (색상으로 그룹 구분)

이번에는:
종(species)에 따른 차이
를 시각화합니다.

#### 종별 히스토그램
#### 코드

```python
sns.histplot(
    data=iris_data,
    x='Petal Length (cm)',
    hue='Species'
)

plt.title("Histogram of Petal Lengths, by Species")
```



#### 설명

* data=
데이터셋 지정

* x=
분석할 컬럼

* hue=
색상 기준 그룹

여기서는:
```python
Species
```
별로 나눔

#### 그래프 해석

각 종마다:
* 꽃잎 길이 분포가 다름

#### 종별 KDE Plot
#### 코드

```python
sns.kdeplot(
    data=iris_data,
    x='Petal Length (cm)',
    hue='Species',
    shade=True
)

plt.title("Distribution of Petal Lengths, by Species")
```



#### 최신 seaborn 권장 코드

```python
sns.kdeplot(
    data=iris_data,
    x='Petal Length (cm)',
    hue='Species',
    fill=True
)
```

#### 그래프 해석

흥미로운 패턴 발견 가능:

#### Iris versicolor 와 Iris virginica

꽃잎 길이가 비슷

#### Iris setosa

완전히 다른 그룹처럼 보임
꽃잎 길이가 훨씬 짧음

#### 매우 중요한 분석 결과

데이터를 보면:
꽃잎 길이만 봐도 Iris setosa를 구분 가능할 수도 있음
