#### Trends (추세)

추세(trend)는 데이터의 변화 패턴을 의미합니다.


> sns.lineplot
* 선 그래프(Line chart)는 시간에 따른 변화 추세를 보여줄 때 가장 적합합니다.
* 여러 개의 선을 사용하면 여러 그룹의 추세를 동시에 비교할 수도 있습니다.

#### Relationship (관계)

데이터 변수들 사이의 관계를 이해하기 위해 다양한 차트 유형을 사용할 수 있습니다.

> sns.barplot
* 막대 그래프는 서로 다른 그룹의 수량을 비교할 때 유용합니다.

> sns.heatmap
* 히트맵은 숫자 테이블에서 색깔 패턴을 통해 관계를 찾을 수 있게 해줍니다.

> sns.scatterplot
* 산점도는 두 개의 연속형 변수 사이의 관계를 보여줍니다.
* 색상을 추가하면 세 번째 범주형 변수와의 관계도 함께 표현할 수 있습니다.

> sns.regplot
* 산점도에 회귀선(regression line)을 추가하여 선형 관계를 더 쉽게 확인할 수 있습니다.

> sns.lmplot
* 여러 색상 그룹이 있을 때 여러 개의 회귀선을 그리는 데 유용합니다.

> sns.swarmplot
* 범주형 변수와 연속형 변수의 관계를 보여주는 범주형 산점도입니다.

#### Distribution (분포)

분포 시각화는 어떤 값들이 나타날 수 있는지와, 각 값이 얼마나 자주 나타나는지를 보여줍니다.

> sns.histplot
* 히스토그램은 하나의 숫자형 변수의 분포를 보여줍니다.

> sns.kdeplot
* KDE 그래프는 숫자형 변수의 부드러운 분포를 추정해서 보여줍니다.
* 2차원 KDE 그래프도 가능합니다.

> sns.jointplot
* 2D KDE 그래프와 각 변수의 개별 KDE 그래프를 동시에 보여줍니다.

#### Seaborn으로 스타일 변경하기

지금까지 사용한 명령어들은 기본적으로 보기 좋은 스타일을 제공했습니다.
하지만 코드 한 줄만 추가하면 그래프 스타일을 쉽게 변경할 수 있습니다.
먼저 이전 튜토리얼에서 사용했던 선 그래프 코드를 다시 살펴봅니다.

```python
# 파일 경로
spotify_filepath = "../input/spotify.csv"

# 데이터 읽기
spotify_data = pd.read_csv(spotify_filepath, index_col="Date", parse_dates=True)

# 선 그래프
plt.figure(figsize=(12,6))
sns.lineplot(data=spotify_data)
```



다음 코드 한 줄로 그래프 테마를 변경할 수 있습니다.

```python
# 스타일을 "dark" 테마로 변경
sns.set_style("dark")

# 선 그래프
plt.figure(figsize=(12,6))
sns.lineplot(data=spotify_data)
```



#### Seaborn에서 사용할 수 있는 5가지 테마

Seaborn은 다음과 같은 다섯 가지 테마를 제공합니다.

* "darkgrid"
* "whitegrid"
* "dark"
* "white"
* "ticks"

원하는 테마 이름만 넣어서 스타일을 바꿀 수 있습니다.

예시:
```python
sns.set_style("whitegrid")
```
다음 실습에서는 이 테마들을 직접 바꿔보며 어떤 스타일이 가장 마음에 드는지 실험하게 됩니다.
