지금까지는 제공된 데이터셋을 사용해왔습니다.
이번 튜토리얼에서는 여러분만의 데이터셋을 사용하는 방법을 배우게 됩니다.
그리고 다음 실습에서는 직접 데이터 시각화를 설계하고 만들어보게 됩니다.

또한, 여러분의 데이터셋을 저장하거나 수만 개의 공개 데이터셋에 빠르게 접근할 수 있는 Kaggle Datasets 도구에 대해 배우게 됩니다.

#### Kaggle Datasets

아래 링크를 통해 Kaggle Datasets에 접근할 수 있습니다.

[Kaggle Datasets](https://www.kaggle.com/datasets)

이 링크를 열면 프로젝트에 사용할 수 있는 다양한 데이터셋 모음을 볼 수 있습니다.

페이지에 보이는 데이터셋들은 스크린샷과 다를 수 있습니다.
왜냐하면 매일 새로운 데이터셋이 계속 업로드되기 때문입니다!

Kaggle Datasets에는 여러 파일 형식이 존재합니다.

예시:
* CSV
* JSON
* SQLite
* BigQuery

이번 과정에서는 지금까지 사용했던 형식인 CSV 파일이 포함된 데이터셋을 선택해서 사용할 예정입니다.

#### 특정 데이터셋 검색하기

화면 상단의 검색창을 사용하면 원하는 데이터셋을 찾을 수 있습니다.

예를 들어:
* 만화 캐릭터(comic book characters) 관련 데이터셋을 찾고 싶다면
* 검색창에 "comic" 을 입력합니다.

그러면 FiveThirtyEight Comic Characters Dataset 을 찾을 수 있습니다.

이 데이터셋에는 총 3개의 파일이 있으며,
그중 CSV 파일도 포함되어 있습니다.

#### 데이터셋 선택하기

데이터셋을 클릭하면 해당 데이터셋 설명 페이지로 이동합니다.
페이지 왼쪽의 Data Explorer 영역을 아래로 스크롤하면 포함된 파일 목록을 볼 수 있습니다.
이 데이터셋에는 다음 세 파일이 포함되어 있습니다.

1. README.md
2. dc-wikia-data.csv
3. marvel-wikia-data.csv

기본적으로 첫 번째 파일이 선택되어 있습니다.

CSV 파일 내용을 미리 보고 싶다면:
* dc-wikia-data.csv 또는 marvel-wikia-data.csv를 클릭하면 됩니다.

지금 시간을 내서 페이지의 다른 탭들도 살펴보세요.

예를 들어:
* Discussion 탭에서는 다른 사람들이 데이터셋에 대해 어떤 이야기를 하는지도 볼 수 있습니다.

#### 나만의 데이터셋 사용하기

다음 실습에서는 원하는 CSV 데이터셋을 직접 선택해서 사용하게 됩니다.
앞서 배운 것처럼 Kaggle Datasets에는 매우 많은 공개 데이터셋이 있습니다.
하지만 자신의 데이터셋을 사용하고 싶다면 먼저 Kaggle Datasets에 업로드해야 합니다.

#### 데이터셋 업로드 방법

1. 다시 Kaggle Datasets 로 이동합니다.
2. 왼쪽 상단의 "New Dataset" 버튼을 클릭합니다.

#### CSV 파일이 아니어도 괜찮아요

CSV 파일 형식에 익숙하지 않아도 괜찮습니다.

Kaggle Datasets는:
* Google 스프레드시트
* Excel 파일(.xlsx)

같은 표 형식 데이터를 자동으로 CSV 파일로 변환해줍니다.
즉, Excel 파일이나 구글 시트를 그대로 업로드해도 자동 처리됩니다.
