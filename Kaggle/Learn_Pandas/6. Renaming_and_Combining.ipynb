#### Renaming (이름 변경)

먼저 소개하는 함수는:
```python
rename()
```
입니다.

이 함수는:
* 인덱스 이름(index)
* 컬럼 이름(column)
을 변경할 수 있습니다.

#### 컬럼 이름 바꾸기

예시:
points 컬럼 이름을 score로 변경:
```python
reviews.rename(columns={'points': 'score'})
```
설명:
* columns= 에 dictionary 전달
* '기존이름': '새이름'

즉:
```python
'points' → 'score'
```
로 변경됩니다.

#### 인덱스 이름 바꾸기

예시:
```python
reviews.rename(index={0: 'firstEntry', 1: 'secondEntry'})
```

설명:
* index 값 자체를 변경

즉:
```python
0 → firstEntry
1 → secondEntry
```

#### 실무 팁

보통은:
* 컬럼 이름 변경은 자주 함
* 인덱스 값 변경은 거의 안 함

인덱스를 바꾸는 경우에는 보통:
```python
set_index()
```
를 더 많이 사용합니다.

#### rename_axis()

행(rows)과 열(columns)에도 이름(name)을 붙일 수 있습니다.

이름을 변경할 때:
```python
rename_axis()
```
를 사용합니다.

예시:
```python
reviews.rename_axis("wines", axis='rows') \
       .rename_axis("fields", axis='columns')
```

설명:
* 행 축 이름 → "wines"
* 열 축 이름 → "fields"

#### Combining (데이터 합치기)

데이터 분석에서는 여러 DataFrame/Series를 합쳐야 하는 경우가 많습니다.
Pandas에는 대표적으로 3가지 방법이 있습니다.

복잡도 순서:
1. concat()
2. join()
3. merge()

튜토리얼에서는:
* concat()
* join()
만 설명합니다.
(merge()는 join으로 대부분 가능)

#### concat()

가장 단순한 합치기 방법입니다.

```python
pd.concat()
```
리스트 형태의 여러 DataFrame/Series를 이어붙입니다.

#### concat 예시

캐나다 유튜브 데이터 + 영국 유튜브 데이터 합치기:
```python
pd.concat([canadian_youtube, british_youtube])
```

설명:
* 컬럼 구조가 같을 때 사용
* 행(row)을 아래로 이어붙임

즉:
```python
캐나다 데이터
+
영국 데이터
=
하나의 큰 DataFrame
```

#### join()

join()은 공통 인덱스를 기준으로 DataFrame을 합칩니다.
concat()보다 조금 더 복잡합니다.

#### join 예시

```python
left = canadian_youtube.set_index(['title', 'trending_date'])
right = british_youtube.set_index(['title', 'trending_date'])

left.join(right, lsuffix='_CAN', rsuffix='_UK')
```

설명:
#### 1단계

두 DataFrame 모두:
```python
title
trending_date
```
를 인덱스로 설정

#### 2단계

공통 인덱스를 기준으로 join

즉:
같은 제목(title) + 같은 날짜(trending_date)
를 가진 데이터를 연결

#### lsuffix / rsuffix

캐나다와 영국 데이터는 같은 컬럼 이름이 많습니다.

예:
* views
* likes
* comments

이런 컬럼이 충돌하므로 suffix 필요

```python
lsuffix='_CAN'
rsuffix='_UK'
```

결과:
```python
views_CAN
views_UK
```
처럼 구분됩니다.
