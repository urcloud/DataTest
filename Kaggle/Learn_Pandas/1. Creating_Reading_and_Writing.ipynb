#### 데이터 만들기 (Creating data)

pandas의 핵심 객체는 두 가지입니다.

1. DataFrame
2. Series

#### DataFrame

DataFrame은 표(table) 형태의 데이터입니다.
각 값(entry)은 특정한 행(row)과 열(column)에 위치합니다.

예시:
```python
pd.DataFrame({'Yes': [50, 21], 'No': [131, 2]})
```

|   | Yes | No  |
| - | --- | --- |
| 0 | 50  | 131 |
| 1 | 21  | 2   |

예를 들어:
* (0, No) 위치 값 → 131
* (0, Yes) 위치 값 → 50

DataFrame에는 숫자뿐 아니라 문자열도 저장할 수 있습니다.
```python
pd.DataFrame({
    'Bob': ['I liked it.', 'It was awful.'],
    'Sue': ['Pretty good.', 'Bland.']
})
```

결과:
|   | Bob           | Sue          |
| - | ------------- | ------------ |
| 0 | I liked it.   | Pretty good. |
| 1 | It was awful. | Bland.       |

#### DataFrame 생성 방식 설명

```python
pd.DataFrame({...})
```
이렇게 dictionary(딕셔너리)를 사용해 생성합니다.

구조는:
```python
{
    '열이름': [값1, 값2, 값3]
}
```
* key → 열(column) 이름
* value → 해당 열의 데이터 목록(list)

#### Index (인덱스)

기본적으로 pandas는 행 번호를 0,1,2... 자동 생성합니다.

하지만 직접 지정할 수도 있습니다.

```python
pd.DataFrame(
    {
        'Bob': ['I liked it.', 'It was awful.'],
        'Sue': ['Pretty good.', 'Bland.']
    },
    index=['Product A', 'Product B']
)
```

결과:
|           | Bob           | Sue          |
| --------- | ------------- | ------------ |
| Product A | I liked it.   | Pretty good. |
| Product B | It was awful. | Bland.       |

여기서:
* Product A
* Product B
가 인덱스(index)입니다.

#### Series

Series는 데이터 값들의 “리스트”입니다.

DataFrame이 표라면,
Series는 한 줄짜리 열(column)이라고 생각하면 됩니다.

예시:
```python
pd.Series([1, 2, 3, 4, 5])
```

결과:
```python
0    1
1    2
2    3
3    4
4    5
dtype: int64
```

Series도 index를 지정할 수 있습니다.
```python
pd.Series(
    [30, 35, 40],
    index=['2015 Sales', '2016 Sales', '2017 Sales'],
    name='Product A'
)
```

결과:
```python
2015 Sales    30
2016 Sales    35
2017 Sales    40
Name: Product A
dtype: int64
```

설명:
* 값: 30,35,40
* index: 연도
* Series 이름(name): "Product A"

#### DataFrame과 Series 관계

DataFrame은 여러 개의 Series를 붙여놓은 구조라고 이해하면 쉽습니다.

즉:
* Series = 한 개의 열(column)
* DataFrame = 여러 Series들의 집합

#### 데이터 파일 읽기 (Reading data files)

실제로는 데이터를 직접 만드는 경우보다,
이미 존재하는 데이터를 읽어오는 경우가 훨씬 많습니다.

#### CSV 파일

가장 기본적인 데이터 파일 형식은 CSV입니다.

CSV 예시:
```python
Product A,Product B,Product C
30,21,9
35,34,1
41,11,11
```

CSV는:
* 값을 쉼표(comma)로 구분
* 그래서 이름이 CSV (Comma-Separated Values)

#### CSV 읽기

pandas에서는 read_csv() 사용:
```python
wine_reviews = pd.read_csv(
    "../input/wine-reviews/winemag-data-130k-v2.csv"
)
```
