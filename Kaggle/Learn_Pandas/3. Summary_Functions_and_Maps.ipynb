#### 요약 함수 (Summary functions)

Pandas는 데이터를 빠르게 요약해주는 여러 함수들을 제공합니다.

예를 들어:
```python
reviews.points.describe()
```
이 코드는 points 컬럼의 통계 요약 정보를 보여줍니다.

출력 예시:
* count: 데이터 개수
* mean: 평균
* std: 표준편차
* min/max: 최소값/최대값
* 25%, 50%, 75%: 분위수
즉, 숫자형 데이터의 전체적인 특징을 한 번에 파악할 수 있습니다.

문자열 컬럼에 describe()를 쓰면 결과가 달라집니다.
```python
reviews.taster_name.describe()
```

문자열 데이터에서는:
* count: 값 개수
* unique: 고유값 개수
* top: 가장 많이 나온 값
* freq: 가장 많이 나온 값의 빈도
등을 보여줍니다.

즉, describe()는 데이터 타입에 따라 알아서 다른 요약 정보를 제공하는 함수입니다.

#### Maps (매핑)

Map은 데이터를 다른 형태로 변환하는 작업입니다.

데이터 분석에서는:
* 기존 값을 수정하거나
* 새로운 형태의 데이터를 만들 때
매우 자주 사용됩니다.

#### map()

map()은 Series(한 컬럼)에 대해 값 하나씩 변환할 때 사용합니다.

예제:
```python
review_points_mean = reviews.points.mean()

reviews.points.map(lambda p: p - review_points_mean)
```

의미:
각 점수에서 평균 점수를 빼라는 뜻입니다.

예:
* 평균이 88.4인데
* 어떤 와인이 87점이면

```python
87 - 88.4 = -1.4
```
이렇게 평균 기준으로 얼마나 높은지/낮은지 계산합니다.

#### lambda 함수

```python
lambda p: p - review_points_mean
```

의미:
* p 하나를 받아서
* 평균을 뺀 값을 반환

즉:
```python
def func(p):
    return p - review_points_mean
```
과 같은 뜻입니다.

#### apply()

apply()는 DataFrame 전체의 행(row)이나 열(column)에 함수 적용할 때 사용합니다.

예제:
```python
def remean_points(row):
    row.points = row.points - review_points_mean
    return row

reviews.apply(remean_points, axis='columns')
```

#### axis='columns'

각 행(row)을 함수에 넣겠다는 의미입니다.

즉:
* 첫 번째 와인 row 전달
* 두 번째 와인 row 전달
* ...
반복 수행됩니다.

#### 함수 동작

```python
row.points = row.points - review_points_mean
```
현재 행의 points 값에서 평균을 빼서 수정합니다.

#### map vs apply 차이

| 함수      | 대상        | 용도        |
| ------- | --------- | --------- |
| map()   | Series    | 값 하나씩 변환  |
| apply() | DataFrame | 행/열 전체 처리 |

#### 원본 데이터는 안 바뀜

중요한 점:
```python
reviews.points.map(...)
```
이나

```python
reviews.apply(...)
```
는 새로운 결과를 반환할 뿐,
원본 reviews 데이터는 수정되지 않습니다.

즉:
```python
reviews.head(1)
```
를 다시 보면 원래 값 그대로입니다.

#### Pandas의 벡터 연산(Vectorized Operation)

사실 map/apply보다 더 빠른 방법이 있습니다.

```python
reviews.points - review_points_mean
```

이 코드만으로도:
```python
reviews.points.map(lambda p: p - mean)
```
과 같은 결과가 나옵니다.

왜냐하면 Pandas는:
* Series 전체에 대해
* 자동으로 연산을 수행하기 때문입니다.
이걸 벡터화 연산(Vectorized operation)이라고 합니다.

#### 문자열 컬럼끼리 합치기

```python
reviews.country + " - " + reviews.region_1
```

결과:
```python
Italy - Etna
France - Alsace
```
처럼 문자열도 Series끼리 합칠 수 있습니다.

#### 벡터 연산 장점

```python
+ - * / > < ==
```

같은 연산자는 Pandas 내부 최적화를 사용하므로:
* map()
* apply()
보다 훨씬 빠릅니다.

#### 하지만 map/apply가 필요한 경우

복잡한 조건문이나 사용자 정의 로직은
벡터 연산만으로 어렵습니다.

예:
```python
if score > 90:
    return "good"
else:
    return "bad"
```

이런 경우에는:
* map()
* apply()
를 사용합니다.
