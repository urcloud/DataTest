#### iloc 문법

```python
df.iloc[행, 열]
```

예:
```python
reviews.iloc[0, 0]
```

#### 전체 행에서 특정 컬럼 가져오기

```python
reviews.iloc[:, 0]
```

여기서:
* : = 전체
* 0 = 첫 번째 컬럼

즉:
모든 행의 첫 번째 컬럼 가져오기

결과:
```python
0      Italy
1   Portugal
```

#### 슬라이싱
#### 처음 3개 행

```python
reviews.iloc[:3, 0]
```

뜻:
* 0~2번째 행
* 첫 번째 컬럼

결과:
```python
0 Italy
1 Portugal
2 US
```

주의:
Python 슬라이싱은 끝 포함 안함.

```python
:3
```
→ 0,1,2

#### 1~2번째 행

```python
reviews.iloc[1:3, 0]
```
→ 1,2 행 선택

#### 여러 행 선택

```python
reviews.iloc[[0,1,2], 0]
```

#### 음수 인덱스

#### loc (라벨 기반 선택)

iloc는 위치 기반.
loc는 이름 기반.

예시
```python
reviews.loc[0, 'country']
```

의미:
* index가 0인 행
* country 컬럼

결과:
```python
Italy
```

#### loc vs iloc 차이
#### iloc

숫자 위치 기준

```python
reviews.iloc[0]
```
→ 첫 번째 행

#### loc

인덱스 이름 기준

```python
reviews.loc[0]
```
→ index 이름이 0인 행

#### 여러 컬럼 선택

```python
reviews.loc[:, ['taster_name', 'points']]
```

뜻:
* 모든 행
* taster_name, points 컬럼만 선택

#### loc와 iloc 슬라이싱 차이 (매우 중요)
#### iloc

끝 포함 안함

```python
reviews.iloc[0:3]
```
→ 0,1,2

#### loc

끝 포함함

```python
reviews.loc[0:3]
```
→ 0,1,2,3

#### 인덱스 변경

#### 조건 필터링 (핵심)
#### 이탈리아 와인만

```python
reviews.loc[reviews.country == 'Italy']
```

뜻:
1. country가 Italy인지 검사
2. True인 행만 선택

#### 조건 여러 개

```python
reviews.loc[
    (reviews.country == 'Italy') &
    (reviews.points >= 90)
]
```

뜻:
* 이탈리아 와인이고
* 점수가 90 이상
