#### Groupwise analysis (그룹별 분석)

우리가 자주 사용했던 함수 중 하나가 value_counts() 입니다.

예를 들어:
```python
reviews.groupby('points').points.count()
```

의 의미는:
1. points 값이 같은 데이터끼리 그룹화
2. 각 그룹에서 points 개수 세기
입니다.

즉, 사실 value_counts()는 내부적으로 groupby() 기반 동작이라고 볼 수 있습니다.

예시: 점수별 와인 개수 세기
```python
reviews.groupby('points').points.count()
```

출력 예시:
```python
points
80     397
81     692
...
99      33
100     19
```

의미:
* 80점 와인: 397개
* 81점 와인: 692개
* 100점 와인: 19개

#### 그룹별 최소값 구하기

각 점수 그룹에서 가장 싼 와인을 찾기:
```python
reviews.groupby('points').price.min()
```

의미:
* 같은 점수끼리 묶고
* 각 그룹에서 price 최소값 구하기

결과:
```python
points
80      5.0
81      5.0
...
100    80.0
```

즉:
* 80점 와인 중 최저가 = 5달러
* 100점 와인 중 최저가 = 80달러

#### apply() 사용하기

apply()를 사용하면 그룹별 DataFrame 자체를 직접 다룰 수 있습니다.

예시:
```python
reviews.groupby('winery').apply(lambda df: df.title.iloc[0])
```

의미:
1. 와이너리별로 그룹화
2. 각 그룹(df)에서 첫 번째 와인 제목 가져오기

#### lambda df 의미

```python
lambda df:
```

여기서 df는:
* 각 그룹의 DataFrame
입니다.

즉:
```python
df.title.iloc[0]
```
은:
* 그 그룹의 첫 번째 행 title 가져오기
라는 뜻입니다.

#### 여러 컬럼으로 그룹화

groupby()는 컬럼 여러 개도 가능합니다.

예시:
```python
reviews.groupby(['country', 'province'])
```

의미:
* country + province 조합별 그룹 생성

#### 국가 + 지역별 최고 점수 와인 찾기

```python
reviews.groupby(['country', 'province']).apply(
    lambda df: df.loc[df.points.idxmax()]
)
```

#### 핵심

```python
df.points.idxmax()
```
→ 가장 점수가 높은 행의 index 반환

```python
df.loc[...]
```
→ 그 행 전체 가져오기

즉:
* 국가 + 지역별 최고 평점 와인 선택

#### agg() 사용하기

agg()는 여러 통계 함수를 한 번에 적용할 수 있습니다.

예시:
```python
reviews.groupby(['country']).price.agg([len, min, max])
```

의미:
국가별로:
* 데이터 개수(len)
* 최소 가격(min)
* 최대 가격(max)
동시에 계산

결과 예시:
```python
country      len   min   max
Argentina   3800   4.0  230.0
```

즉:
* 아르헨티나 와인 3800개
* 최저가 4달러
* 최고가 230달러

#### Multi-index (다중 인덱스)

일반 DataFrame은 보통 인덱스가 하나입니다.

예:
```python
0
1
2
```

하지만 groupby() 결과는 여러 단계의 인덱스를 가질 수 있습니다.

예시:
```python
countries_reviewed = reviews.groupby(
    ['country', 'province']
).description.agg([len])
```

결과:
```python
country     province
Argentina   Mendoza Province
```
처럼 인덱스가 두 단계입니다.
이걸 MultiIndex라고 합니다.

#### reset_index()

MultiIndex를 일반 컬럼 형태로 되돌리는 함수:
```python
countries_reviewed.reset_index()
```

결과:
```python
country   province   len
```

즉:
* 인덱스를 다시 일반 컬럼으로 변환
실무에서 매우 자주 사용합니다.

#### Sorting (정렬)

groupby() 결과는 기본적으로:
* 값 기준이 아니라
* 인덱스 기준
정렬입니다.

원하는 순서로 보려면 직접 정렬해야 합니다.

#### sort_values()

값 기준 정렬:
```python
countries_reviewed.sort_values(by='len')
```

의미:
* len 컬럼 기준 오름차순 정렬

#### 내림차순 정렬

```python
countries_reviewed.sort_values(
    by='len',
    ascending=False
)
```

의미:
* 큰 값부터 정렬
실무에서 더 자주 사용합니다.

#### sort_index()

인덱스 기준 정렬:
```python
countries_reviewed.sort_index()
```

의미:
* 인덱스(country, province) 기준 정렬

#### 여러 컬럼 기준 정렬

```python
countries_reviewed.sort_values(
    by=['country', 'len']
)
```

의미:
1. country 기준 정렬
2. 같은 country 안에서는 len 기준 정렬
