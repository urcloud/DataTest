#### Dtypes (데이터 타입)

DataFrame의 각 열(column)이나 Series가 가지는 데이터 타입을 dtype이라고 합니다.

예를 들어 reviews 데이터프레임의 price 컬럼 타입을 확인하려면:
```python
reviews.price.dtype
```

결과:
```python
dtype('float64')
```
즉, price 컬럼은 float64 타입(실수형)이라는 뜻입니다.

#### 모든 컬럼의 타입 확인

```python
reviews.dtypes
```

결과:
```python
country        object
description    object
...
variety        object
winery         object
```

#### dtype 의미

Pandas는 데이터를 내부적으로 어떤 형식으로 저장할지 dtype으로 관리합니다.

예시:
* float64 → 64비트 실수
* int64 → 64비트 정수
* object → 문자열(string) 또는 일반 객체

주의할 점:
문자열만 있는 컬럼도 별도의 string 타입이 아니라 보통 object 타입으로 저장됩니다.

#### 타입 변환 (astype)

astype() 함수를 사용하면 컬럼 타입을 변경할 수 있습니다.

예시:
```python
reviews.points.astype('float64')
```
기존 int64였던 points 컬럼을 float64로 변환합니다.

결과:
```python
0         87.0
1         87.0
...
```
정수가 실수 형태(87 → 87.0)로 바뀝니다.

#### 인덱스(index)도 dtype이 있음

```python
reviews.index.dtype
```

결과:
```python
dtype('int64')
```
즉, 인덱스도 데이터 타입을 가집니다.

#### 기타 데이터 타입

Pandas는 일반 숫자/문자열 외에도:
* categorical data (범주형 데이터)
* timeseries data (시계열 데이터)
같은 특수 타입도 지원합니다.

다만 여기서는 다루지 않습니다.

#### Missing Data (결측값)

값이 비어 있는 데이터는 NaN으로 표시됩니다.

NaN은:
* Not a Number의 약자입니다.
* 그리고 기술적인 이유로 항상 float64 타입입니다.

#### 결측값 찾기

Pandas에서는:
```python
pd.isnull()
```
또는

```python
pd.notnull()
```
을 사용합니다.

예시:
```python
reviews[pd.isnull(reviews.country)]
```
→ country 값이 비어있는 행만 선택합니다.

#### 결측값 채우기 (fillna)

결측값을 다른 값으로 채우는 작업은 매우 자주 사용됩니다.

Pandas에서는:
```python
fillna()
```
를 사용합니다.

예시:
```python
reviews.region_2.fillna("Unknown")
```
→ NaN 값을 "Unknown"으로 변경합니다.

결과:
```python
0         Unknown
1         Unknown
...
```

#### Backfill 전략

결측값을:
* 뒤쪽에 나오는 첫 번째 정상값(non-null value)으로 채울 수도 있습니다.
* 이 방법을 backfill 전략이라고 합니다.

#### 값 치환하기 (replace)

특정 값을 다른 값으로 바꾸고 싶을 때는:
```python
replace()
```
를 사용합니다.

예시:
```python
reviews.taster_twitter_handle.replace("@kerinokeefe", "@kerino")
```
→ "@kerinokeefe" 를 "@kerino" 로 변경합니다.

#### replace()가 유용한 이유

실제 데이터셋에서는 결측값 대신:
* "Unknown"
* "Undisclosed"
* "Invalid"

같은 문자열이 들어있는 경우가 많습니다.
이럴 때 replace()를 사용하면 쉽게 바꿀 수 있습니다.
