#### 7번

와인 점수를 별점으로 변환하세요.

규칙:
* ⭐⭐⭐ 3 stars
    * 점수 95 이상
    * 또는 Canada 와인

* ⭐⭐ 2 stars
    * 85 이상 95 미만
    
* ⭐ 1 star
    * 나머지

```python
def stars(row):
    if row.country == 'Canada':
        return 3
    elif row.points >= 95:
        return 3
    elif row.points >= 85:
        return 2
    else:
        return 1

star_ratings = reviews.apply(stars, axis='columns')

# Check your answer
# q7.check()
```
