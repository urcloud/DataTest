#### 3번

와인 종류(variety)별 최소/최대 가격은?

index는 variety

컬럼은:
* min
* max
가 되도록 DataFrame 생성.

````python
price_extremes = reviews.groupby(
    'variety'
).price.agg([min, max])

# Check your answer
# q3.check()
````

#### 4번

가장 비싼 와인 종류는 무엇인가?

3번에서 만든 DataFrame을:
* minimum price 기준 내림차순
* 같으면 maximum price 기준 내림차순
정렬하시오.

````python
sorted_varieties = price_extremes.sort_values(by=['min', 'max'], ascending=False)

# Check your answer
# q4.check()
````
