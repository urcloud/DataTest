#### 1번

많은 프로그래밍 언어에는 sign이라는 내장 함수가 있습니다.
Python에는 없지만, 직접 만들 수 있습니다!

아래 셀에서 sign이라는 함수를 정의하세요.

* 숫자가 음수이면 -1
* 양수이면 1
* 0이면 0

을 반환하도록 작성하면 됩니다.

````python
def sign(n):
    if n<0:
        return -1
    elif n>0:
        return 1
    else:
        return 0

# Check your answer
# q1.check()
````

#### 2번

이전 연습문제의 to_smash 함수에 "로그(logging)"를 추가하기로 했습니다.

함수를 수정해서:

* 사탕이 1개면 "candy"
* 그 외에는 "candies"

가 출력되도록 바꾸세요.

````python
def to_smash(total_candies):
    """Return the number of leftover candies that must be smashed after distributing
    the given number of candies evenly between 3 friends.
    
    >>> to_smash(91)
    1
    """

    if total_candies == 1:
        num = "candy"
    else:
        num = "candies"

    print("Splitting", total_candies, num)
    return total_candies % 3
````

#### 4번

아래 is_negative 함수는 올바르게 동작합니다.
하지만 너무 길게 작성되어 있습니다.

동일한 기능을 하는 더 짧은 버전(concise_is_negative)을 만들어보세요.

힌트:
* 단 한 줄로도 가능합니다.
* ternary(a if condition else b)도 필요 없습니다.

````python
def is_negative(number):
    if number < 0:
        return True
    else:
        return False

def concise_is_negative(number):
    return bool(number < 0)
````

우리는 bool()을 정수에 사용하면:
* 0 → False
* 그 외 → True

라는 것을 배웠습니다.

그렇다면 반대로 int(True)와 int(False)는 어떻게 될까요?

이걸 활용해서 아래 함수를 짧게 작성해보세요.

```python
def exactly_one_topping(ketchup, mustard, onion):
    """세 가지 토핑 중 정확히 하나만 원하는지 반환"""
```

즉:
딱 하나만 True이면 True

````python
def exactly_one_topping(ketchup, mustard, onion):
    """Return whether the customer wants exactly one of the three available toppings
    on their hot dog.
    """
    return ((bool(ketchup)+bool(mustard)+bool(onion))==1)

# Check your answer
# q6.check()
````

#### 7번 🌶️ (선택 문제)

이번에는 간단한 블랙잭(21) 게임을 다룹니다.

게임 규칙 요약:
* 진행 방식
* 플레이어는 카드 2장을 받음
* 딜러는 카드 1장을 공개함
* 플레이어는 원하는 만큼 카드를 더 받을 수 있음(hit)
* 합이 21 초과면 즉시 패배

그 후 딜러는:
* 합이 17 이상이 될 때까지 카드 추가
* 21 초과면 플레이어 승리
* 그렇지 않으면 더 높은 숫자가 승리
* 동점은 딜러 승리

카드 계산
* J/Q/K = 10
* A(에이스)는 1 또는 11 가능

예:
* A + 8 = 19
* A + 8 + 8 = 17

해야 할 일

아래 함수는 플레이어가 카드를 더 받을지 결정하는 함수입니다.

```python
def should_hit(dealer_total, player_total, player_low_aces, player_high_aces):
    return False
```

현재는 무조건 False라서:
절대 카드를 더 받지 않는 매우 멍청한 전략입니다.
이 함수를 더 똑똑하게 수정해서 승률을 높여보세요.

````python
def should_hit(dealer_total, player_total, player_low_aces, player_high_aces):
    """Return True if the player should hit (request another card) given the current game
    state, or False if the player should stay.
    When calculating a hand's total value, we count aces as "high" (with value 11) if doing so
    doesn't bring the total above 21, otherwise we count them as low (with value 1). 
    For example, if the player's hand is {A, A, A, 7}, we will count it as 11 + 1 + 1 + 7,
    and therefore set player_total=20, player_low_aces=2, player_high_aces=1.
    """

    # 11 이하면 무조건 hit
    if player_total <= 11:
        return True

    # 딜러가 강한 카드(7 이상)를 가지고 있고
    # 내가 16 이하라면 hit
    if dealer_total >= 7 and player_total <= 16:
        return True

    # 내가 12 이하라면 hit
    if player_total <= 12:
        return True

    return False
````
