Luigi는 Mario Kart 레이스에서 어떤 아이템이 승리에 가장 도움이 되는지 분석하려고 합니다.
그는 다음과 같은 형태의 데이터(딕셔너리 리스트)를 가지고 있습니다.

데이터 의미
* 'items'
    → 레이서가 경기 중 획득한 아이템 리스트
* 'finish'
    → 경기 순위

Luigi는 아래 함수를 작성했습니다.
이 함수는:
"1등을 한 플레이어들이 획득한 각 아이템의 등장 횟수"
를 딕셔너리 형태로 반환합니다.

작은 예제 데이터에서는 정상 동작했습니다.
하지만 전체 데이터셋에서는 TypeError가 발생하며 프로그램이 종료되었습니다.

* 왜 오류가 발생하는지 찾아보세요.
* 아래 코드를 수정해서 에러 없이 동작하게 만드세요.

힌트: 이 버그는 튜토리얼에서 star import(*)를 설명할 때 등장했던 문제와 비슷합니다.

````python
def best_items(racers):
    """Given a list of racer dictionaries, return a dictionary mapping items to the number
    of times those items were picked up by racers who finished in first place.
    """
    winner_item_counts = {}
    for i in range(len(racers)):
        # The i'th racer dictionary
        racer = racers[i]
        # We're only interested in racers who finished in first
        if racer['finish'] == 1:
            for i in racer['items']:
                # Add one to the count for this item (adding it to the dict if necessary)
                if i not in winner_item_counts:
                    winner_item_counts[i] = 0
                winner_item_counts[i] += 1

        # Data quality issues :/ Print a warning about racers with no name set. We'll take care of it later.
        if racer['name'] is None:
            print("WARNING: Encountered racer with unknown name on iteration {}/{} (racer = {})".format(
                i+1, len(racers), racer['name'])
                 )
    return winner_item_counts

sample = [
    {'name': 'Peach', 'items': ['green shell', 'banana', 'green shell',], 'finish': 3},
    {'name': 'Bowser', 'items': ['green shell',], 'finish': 1},
    {'name': None, 'items': ['mushroom',], 'finish': 2},
    {'name': 'Toad', 'items': ['green shell', 'mushroom'], 'finish': 1},
]
best_items(sample)

from learntools.python.luigi_analysis import full_dataset

def best_items(racers):
    winner_item_counts = {}

    for i in range(len(racers)):
        racer = racers[i]

        # 1등 선수만 처리
        if racer['finish'] == 1:

            # 변수 이름 변경!
            for item in racer['items']:

                if item not in winner_item_counts:
                    winner_item_counts[item] = 0

                winner_item_counts[item] += 1

        # 이름 없는 선수 경고
        if racer['name'] is None:
            print(
                "WARNING: Encountered racer with unknown name on iteration {}/{} (racer = {})".format(
                    i + 1,
                    len(racers),
                    racer['name']
                )
            )

    return winner_item_counts

# 실행
best_items(full_dataset)
````

블랙잭 손패(Blackjack hand)를 표현하는 새로운 타입을 만든다고 가정해봅시다.
우리가 원하는 기능 중 하나는:

```python
>
<=
```
같은 비교 연산자를 직접 정의하는 것입니다.

예를 들면:
```python
hand1 = BlackjackHand(['K', 'A'])
hand2 = BlackjackHand(['7', '10', 'A'])

hand1 > hand2
```
가 True가 되도록 만들고 싶습니다.

이번 문제에서는 실제 클래스를 만들지는 않지만, 아래 함수에 작성할 코드가 실제로는 __gt__ 메서드에 들어갈 코드와 매우 유사합니다.

* blackjack_hand_greater_than 함수를 docstring 설명에 맞게 구현하세요.

```python
def blackjack_hand_greater_than(hand_1, hand_2):
```

hand_1이 hand_2를 이기려면 다음 조건을 만족해야 합니다.
1. hand_1의 총합이 21을 초과하면 안 된다.
2. 그리고 아래 둘 중 하나를 만족해야 한다.
    * hand_1의 총합 > hand_2의 총합
    * 또는 hand_2가 21을 초과했다.

카드 표현 방식:
손패는 카드 문자열의 리스트로 표현됩니다.

예시:
```python
['K', 'A']
```

카드 점수 규칙:
* 숫자 카드 → 숫자 그대로
* 'J', 'Q', 'K' → 10점
* 'A' → 1점 또는 11점 가능

A(에이스) 계산 규칙

에이스는:
"21을 넘지 않는 선에서 최대 점수가 되도록"
계산해야 합니다.

예시:
```python
['A', 'A', '9'] → 21
```
(11 + 1 + 9)

```python
['A', 'A', '9', '3'] → 14
```
(1 + 1 + 9 + 3)

````python
def hand_total(hand):
    total = 0
    aces = 0

    for card in hand:

        # 숫자 카드
        if card.isdigit():
            total += int(card)

        # J, Q, K
        elif card in ['J', 'Q', 'K']:
            total += 10

        # A
        elif card == 'A':
            total += 1
            aces += 1

    # 에이스를 11로 업그레이드 가능한지 확인
    for _ in range(aces):
        if total + 10 <= 21:
            total += 10

    return total


def blackjack_hand_greater_than(hand_1, hand_2):

    total1 = hand_total(hand_1)
    total2 = hand_total(hand_2)

    # hand_1이 bust면 무조건 패배
    if total1 > 21:
        return False

    # hand_2가 bust면 승리
    if total2 > 21:
        return True

    # 일반 비교
    return total1 > total2
````
