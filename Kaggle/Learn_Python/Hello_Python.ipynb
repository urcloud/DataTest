### Hello, Python! 해석

파이썬(Python)은 영국의 코미디 그룹인 Monty Python의 이름을 따서 지어졌습니다. 그래서 첫 번째 파이썬 프로그램도 그들의 “Spam” 스케치를 오마주합니다.

먼저 아래 코드를 읽고 실행하면 어떤 결과가 나올지 예측해보세요.
(전혀 감이 안 와도 괜찮습니다!)

````python
spam_amount = 0
print(spam_amount)

# Ordering Spam, egg, Spam, Spam, bacon and Spam (4 more servings of Spam)
spam_amount = spam_amount + 4

if spam_amount > 0:
    print("But I don't want ANY spam!")

viking_song = "Spam " * spam_amount
print(viking_song)
````

출력 결과:

0

But I don't want ANY spam!

Spam Spam Spam Spam

### 코드 설명
#### 1. 변수 할당 (Variable assignment)

```python
spam_amount = 0
```

spam_amount라는 변수를 만들고 값 0을 저장합니다.

= 는 할당 연산자(assignment operator) 입니다.

다른 언어(Java, C++)와 달리 파이썬은:
* 변수 선언을 따로 하지 않아도 되고
* 변수 타입(int, string 등)을 미리 지정하지 않아도 됩니다.


#### 2. 함수 호출 (Function calls)

```python
print(spam_amount)
```

출력: 

0

print()는 값을 화면에 출력하는 함수입니다.

함수는:

```python
함수이름(인자)
```
형태로 사용합니다.


#### 3. 주석(Comment)과 재할당(Reassignment)

```python
# Ordering Spam, egg, Spam, Spam, bacon and Spam
```

`#` 으로 시작하면 주석입니다.
주석은 코드 실행에 영향을 주지 않습니다.

```python
spam_amount = spam_amount + 4
```
기존 변수 값을 다시 변경하는 것을 재할당(reassignment) 이라고 합니다.

현재:
```python
spam_amount = 0
```
이므로:

```python
0 + 4 = 4
```

결과적으로:
```python
spam_amount = 4
```
가 됩니다.


#### 조건문(if)

```python
if spam_amount > 0:
    print("But I don't want ANY spam!")
```
의미:

만약 spam_amount가 0보다 크다면
"But I don't want ANY spam!" 출력

현재 값은 4이므로 출력됩니다.


#### 들여쓰기(Indentation)

파이썬은 들여쓰기로 코드 블록을 구분합니다.

```python
if 조건:
    실행할 코드
```
콜론(:) 다음 들여쓰기된 부분이 if문에 속합니다.

다른 언어의 { } 대신 들여쓰기를 사용한다고 생각하면 됩니다.

#### 산술 연산자

| 연산자      | 이름                 | 설명        |
| -------- | ------------------ | --------- |
| `a + b`  | 덧셈                 | a와 b의 합   |
| `a - b`  | 뺄셈                 | a와 b의 차   |
| `a * b`  | 곱셈                 | a와 b의 곱   |
| `a / b`  | 나눗셈(True division) | 일반 나눗셈    |
| `a // b` | 몫(Floor division)  | 소수점 버린 몫  |
| `a % b`  | 나머지(Modulus)       | 나눗셈 후 나머지 |
| `a ** b` | 거듭제곱               | a의 b제곱    |
| `-a`     | 음수화                | a의 음수     |


#### / 와 // 차이
#### 일반 나눗셈 /

```python
print(5 / 2)
```

결과:
```python
2.5
```
항상 float(실수) 결과가 나옵니다.

#### 몫 나눗셈 //

```python
print(5 // 2)
```

결과:
```python
2
```
소수점을 버립니다.
