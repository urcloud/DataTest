#### 함수(Function)란?

우리는 이미 print() 나 abs() 같은 함수를 사용해봤어.

예시:
```python
print("Hello")
abs(-5)
```

하지만 Python에는 훨씬 많은 함수가 있고,
직접 함수를 만드는 것도 매우 중요해.

#### 1. help() 함수 — 함수 설명 보기

```python
help(round)
```
이렇게 입력하면 round() 함수의 사용법 설명이 나와.

출력 예시:
```python
round(number, ndigits=None)
```

의미:
* number → 반올림할 숫자
* ndigits → 몇 자리까지 반올림할지 (선택사항)

예:
```python
round(3.14159, 2)
# 결과: 3.14
```

#### help()의 중요한 점

❌ 잘못된 사용:
```python
help(round(-2.01))
```
이건 round(-2.01)의 결과인 -2에 대한 도움말을 보는 거야.

즉:
1. 먼저 round(-2.01) 계산
2. 결과 -2
3. help(-2) 실행
그래서 이상한 설명이 나옴.

#### 3. 함수 직접 만들기

```python
def least_difference(a, b, c):
```
* def → 함수 정의 시작
* 함수 이름: least_difference
* 입력값(매개변수): a, b, c

전체 코드:
```python
def least_difference(a, b, c):
    diff1 = abs(a - b)
    diff2 = abs(b - c)
    diff3 = abs(a - c)
    return min(diff1, diff2, diff3)
```

의미:
세 숫자 중 가장 차이가 작은 값을 반환함.

예시
```python
least_difference(1, 10, 100)
```

계산:
```python
|1 - 10| = 9
|10 - 100| = 90
|1 - 100| = 99
```

가장 작은 값:
```python
9
```

#### 5. Docstring (문서 설명)

함수 설명을 함수 안에 작성할 수 있음.

```python
def least_difference(a, b, c):
    """
    세 숫자 중 가장 작은 차이를 반환한다.
    """
```
이걸 Docstring 이라고 함.

#### 장점
```python
help(least_difference)
```
하면 설명이 같이 나옴.

좋은 개발자는 대부분 docstring을 작성함.

#### 8. 함수를 다른 함수에 전달하기

Python에서는 함수를 변수처럼 사용할 수 있음.

```python
def mult_by_five(x):
    return 5 * x
def call(fn, arg):
    return fn(arg)
```

의미:
* fn = 함수
* arg = 값

즉:
```python
call(mult_by_five, 1)
```

↓

```python
mult_by_five(1)
```

↓

```python
5
```

#### 9. Higher-order Function (고차 함수)

함수를 입력으로 받는 함수.

예:
```python
max(..., key=함수)
```

예시
```python
def mod_5(x):
    return x % 5
max(100, 51, 14, key=mod_5)
```

각 숫자를 5로 나눈 나머지:
| 숫자  | 나머지 |
| --- | --- |
| 100 | 0   |
| 51  | 1   |
| 14  | 4   |
가장 큰 나머지는 4 → 14 반환

결과:
```python
14
```
