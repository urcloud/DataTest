디버깅(debugging)이 운이 필요한 일처럼 느껴진 적 있나요?
다음 프로그램에는 버그가 있습니다. 버그를 찾아 수정해보세요.

* 리스트 nums 안에
* 7로 나누어 떨어지는 숫자가 하나라도 있으면 True
* 없으면 False

````python
# return은 함수 실행을 즉시 종료시킨다는 점을 기억하세요.
# 그래서 원래 구현은 반복문을 단 한 번만 실행했습니다.

# 우리는 리스트의 모든 원소를 확인한 뒤에야 False를 반환할 수 있습니다.
# (즉, 어떤 숫자도 lucky number가 아니라는 것을 전부 확인해야 함)

# 반면에, True인 경우에는 중간에 바로 반환(return)해도 됩니다.

def has_lucky_number(nums):
    """Return whether the given list of numbers is lucky. 
    A lucky list contains at least one number divisible by 7.
    """
    for num in nums:
        if num % 7 == 0:
            return True
    
    return False
````

다음 Python 표현식의 결과를 예상해보세요.

```python
[1, 2, 3, 4] > 2
```

파이썬 기본 리스트는 numpy/pandas처럼 요소별 비교를 하지 않습니다.

* 리스트의 각 원소가 thresh보다 큰지 비교한 결과를
* True/False 리스트로 반환하는 함수를 구현하세요.

````python
def elementwise_greater_than(L, thresh):
    """Return a list with the same length as L, where the value at index i is 
    True if L[i] is greater than thresh, and False otherwise.
    
    >>> elementwise_greater_than([1, 2, 3, 4], 2)
    [False, False, True, True]
    """

    R = []
    
    for x in L:
        if x > thresh:
            R.append(True)
        else:
            R.append(False)

    return R

# Check your answer
# q2.check()
````

#### 3번

식사 목록이 주어졌을 때:
* 같은 음식이 연속된 날짜에 나온 적 있으면 True
* 아니면 False

````python
def menu_is_boring(meals):
    """Given a list of meals served over some period of time, return True if the
    same meal has ever been served two days in a row, and False otherwise.
    """
    for i in range(len(meals) - 1):
        if meals[i] == meals[i + 1]:
            return True
    
    return False

# Check your answer
# q3.check()
````

슬롯머신 함수 play_slot_machine()은:
* 보통 0 반환
* 가끔 큰 돈 당첨

하지만 플레이할 때마다 $1 비용이 듭니다.

목표:
슬롯머신을 여러 번 실행해서
평균적으로 한 번 플레이할 때 얼마를 벌거나 잃는지 계산하세요.
이 방법을 Monte Carlo Method(몬테카를로 시뮬레이션) 라고 합니다.

````python
# 슬롯머신 함수
play_slot_machine()

def estimate_average_slot_payout(n_runs):
    """Run the slot machine n_runs times and return the average net profit per run.
    Example calls (note that return value is nondeterministic!):
    >>> estimate_average_slot_payout(1)
    -1
    >>> estimate_average_slot_payout(1)
    0.5
    """

    total = 0
    
    for i in range(n_runs):
        total += play_slot_machine() - 1
    
    return total / n_runs
````
