#### 문자열(Strings)

파이썬이 특히 강력한 부분 중 하나는 문자열 처리입니다.
이 섹션에서는 문자열 관련 기본 메서드와 포맷팅 기능을 설명합니다.
데이터 분석(Data Science) 작업에서도 문자열 조작은 매우 자주 사용됩니다.

#### 문자열 문법(String syntax)

파이썬 문자열은 작은따옴표(')나 큰따옴표(") 둘 다 사용할 수 있습니다.

```python
x = 'Pluto is a planet'
y = "Pluto is a planet"

x == y
# True
```

둘은 기능적으로 완전히 같습니다.

#### 따옴표 안에 따옴표 넣기

문자열 안에 작은따옴표가 있을 경우 큰따옴표로 감싸면 편합니다.

```python
print("Pluto's a planet!")
print('My dog is named "Pluto"')
```

출력:
```python
Pluto's a planet!
My dog is named "Pluto"
```

오류 발생 예시
```python
'Pluto's a planet!'
```
파이썬은 어디서 문자열이 끝나는지 헷갈려서 오류가 납니다.

#### Escape 문자 ()

백슬래시(\)를 사용하면 특수문자를 문자열 안에 넣을 수 있습니다.

```python
'Pluto\'s a planet!'
```

출력:
```python
"Pluto's a planet!"
```

#### 자주 사용하는 Escape 문자

| 입력   | 실제 결과     |
| ---- | --------- |
| `\'` | 작은따옴표 `'` |
| `\"` | 큰따옴표 `"`  |
| `\\` | 백슬래시 `\`  |
| `\n` | 줄바꿈       |

예시:
```python
hello = "hello\nworld"
print(hello)
```

출력:
```python
hello
world
```

#### 여러 줄 문자열 (Triple Quotes)

""" """ 를 사용하면 줄바꿈을 그대로 포함할 수 있습니다.

```python
triplequoted_hello = """hello
world"""

print(triplequoted_hello)
```

출력:
```python
hello
world
```

#### print()의 end 옵션

기본적으로 print()는 출력 후 줄바꿈(\n)을 합니다.

```python
print("hello")
print("world")
```

출력:
```python
hello
world
```

줄바꿈 없이 출력하려면:
```python
print("hello", end='')
print("pluto", end='')
```

출력:
```python
hellopluto
```

#### 문자열은 시퀀스(sequence)

문자열은 문자들의 나열입니다.
리스트처럼 인덱싱과 슬라이싱이 가능합니다.

```python
planet = 'Pluto'

planet[0]
# 'P'

planet[-3:]
# 'uto'

len(planet)
# 5
```

반복문도 가능합니다.
```python
[char + '! ' for char in planet]
```

출력:
```python
['P! ', 'l! ', 'u! ', 't! ', 'o! ']
```

#### 문자열은 immutable (변경 불가능)

문자열은 수정할 수 없습니다.

```python
planet[0] = 'B'
```

오류 발생:
```python
TypeError: 'str' object does not support item assignment
```

#### 문자열 메서드(String methods)
#### 대문자 / 소문자 변환

```python
claim = "Pluto is a planet!"

claim.upper()
# 'PLUTO IS A PLANET!'

claim.lower()
# 'pluto is a planet!'
```

#### 문자열 검색

```python
claim.index('plan')
# 11
```

#### 시작/끝 확인

```python
claim.startswith(planet)
# True

claim.endswith('planet')
# False
```
느낌표(!)가 없어서 False입니다.

#### split() 과 join()
#### split()

문자열을 리스트로 나눔

```python
words = claim.split()
```

결과:
```python
['Pluto', 'is', 'a', 'planet!']
```

특정 문자 기준으로 나누기:
```python
datestr = '1956-01-31'

year, month, day = datestr.split('-')
```

#### join()

리스트를 문자열로 합침

```python
'/'.join([month, day, year])
```

결과:
```python
'01/31/1956'
```

#### 유니코드 문자 사용 가능

```python
' 👏 '.join([word.upper() for word in words])
```

결과:
```python
'PLUTO 👏 IS 👏 A 👏 PLANET!'
```

#### 문자열 연결 (+)

```python
planet + ', we miss you.'
```

결과:
```python
'Pluto, we miss you.'
```

#### 숫자와 문자열 연결 시 오류

```python
position = 9

planet + ", you'll always be the " + position + "th planet to me."
```

오류 발생:
```python
TypeError
```

문자열로 변환 필요:
```python
planet + ", you'll always be the " + str(position) + "th planet to me."
```

#### format() 사용

더 깔끔한 방법:
```python
"{}, you'll always be the {}th planet to me.".format(planet, position)
```

결과:
```python
"Pluto, you'll always be the 9th planet to me."
```
{} 부분에 값이 들어갑니다.

#### format() 고급 사용

```python
pluto_mass = 1.303 * 10**22
earth_mass = 5.9722 * 10**24
population = 52910390
```
```python
"{} weighs about {:.2} kilograms ({:.3%} of Earth's mass). It is home to {:,} Plutonians.".format(
    planet,
    pluto_mass,
    pluto_mass / earth_mass,
    population
)
```

출력:
```python
'Pluto weighs about 1.3e+22 kilograms (0.218% of Earth's mass). It is home to 52,910,390 Plutonians.'
```

의미:
* {:.2} → 소수점 2자리
* {:.3%} → 퍼센트 표시
* {: ,} → 천 단위 콤마

#### format() 인덱스 사용

```python
s = """Pluto's a {0}.
No, it's a {1}.
{0}!
{1}!""".format('planet', 'dwarf planet')
```

출력:
```python
Pluto's a planet.
No, it's a dwarf planet.
planet!
dwarf planet!
```

#### 딕셔너리(Dictionary)

딕셔너리는 key-value 구조입니다.

```python
numbers = {'one':1, 'two':2, 'three':3}
```

* key: 'one'
* value: 1

#### 값 접근

```python
numbers['one']
# 1
```

#### 값 추가

```python
numbers['eleven'] = 11
```

결과:
```python
{'one': 1, 'two': 2, 'three': 3, 'eleven': 11}
```

#### 값 수정

```python
numbers['one'] = 'Pluto'
```

결과:
```python
{'one': 'Pluto', 'two': 2, 'three': 3, 'eleven': 11}
```

#### Dictionary Comprehension

```python
planets = ['Mercury', 'Venus', 'Earth']

planet_to_initial = {
    planet: planet[0]
    for planet in planets
}
```

결과:
```python
{
 'Mercury': 'M',
 'Venus': 'V',
 'Earth': 'E'
}
```

#### in 연산자

키 존재 여부 확인

```python
'Saturn' in planet_to_initial
# True
```

#### 딕셔너리 반복문

```python
for k in numbers:
    print("{} = {}".format(k, numbers[k]))
```

출력:
```python
one = Pluto
two = 2
three = 3
eleven = 11
```

#### keys(), values()

```python
planet_to_initial.keys()
planet_to_initial.values()
```

* keys() → 키 목록
* values() → 값 목록

#### items()

키와 값을 동시에 가져오기

```python
for planet, initial in planet_to_initial.items():
    print(planet, initial)
```
