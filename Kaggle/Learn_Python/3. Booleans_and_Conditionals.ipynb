#### Boolean (불리언)

Python에는 bool이라는 자료형이 있습니다.
가질 수 있는 값은 딱 두 개입니다.
* True (참)
* False (거짓)

```python
x = True
print(x)
print(type(x))
```

출력:
```python
True
<class 'bool'>
```
보통은 직접 True, False를 쓰기보다 비교 연산자를 통해 불리언 값을 얻습니다.

#### 비교 연산 (Comparison Operations)

| 연산자      | 의미            |
| -------- | ------------- |
| `a == b` | a와 b가 같다      |
| `a != b` | a와 b가 다르다     |
| `a < b`  | a가 b보다 작다     |
| `a > b`  | a가 b보다 크다     |
| `a <= b` | a가 b보다 작거나 같다 |
| `a >= b` | a가 b보다 크거나 같다 |

예제:
```python
def can_run_for_president(age):
    """미국 대통령 선거에 출마 가능한가?"""
    return age >= 35
```
미국 헌법상 대통령 출마는 최소 35세 이상이어야 하므로:

```python
print(can_run_for_president(19))
```
→ False

```python
print(can_run_for_president(45))
```
→ True

#### 비교 연산 주의점

```python
3.0 == 3
```
→ True

숫자끼리는 타입이 달라도 값이 같으면 참으로 판단합니다.

하지만:
```python
'3' == 3
```
→ False

문자열 '3' 과 숫자 3 은 서로 다른 타입이기 때문입니다.

#### 홀수 판별 예제

```python
def is_odd(n):
    return (n % 2) == 1
```
% 는 나머지 연산자입니다.

예:
```python
5 % 2
```
→ 1

즉 홀수면 나머지가 1이므로 홀수 판별 가능.

매우 중요합니다.

#### 비교

→ "n이 2인가?" 를 묻는 것

#### Boolean 조합하기

Python에서는 다음 키워드로 논리 연산을 합니다.
| 연산    | 의미  |
| ----- | --- |
| `and` | 그리고 |
| `or`  | 또는  |
| `not` | 반대  |

예제:
```python
def can_run_for_president(age, is_natural_born_citizen):
    return is_natural_born_citizen and (age >= 35)
```

조건:
* 미국 태생 시민이어야 하고
* 나이가 35세 이상이어야 함

둘 다 만족해야 하므로 and

#### 괄호를 적극적으로 사용하자

예시:
```python
prepared_for_weather = have_umbrella or rain_level < 5 and have_hood
```
이런 코드는 읽기 어렵고 버그 발생 가능성이 큼.

#### 조건문 (Conditionals)

Python 조건문 키워드:
* if
* elif
* else

예제:
```python
def inspect(x):
    if x == 0:
        print("zero")
    elif x > 0:
        print("positive")
    elif x < 0:
        print("negative")
    else:
        print("unknown")
```

동작:
* 0이면 → zero
* 양수면 → positive
* 음수면 → negative
* 그 외 → unknown

#### bool() 변환

Python에는 bool() 함수가 있습니다.
```python
print(bool(1))
```
→ True
```python
print(bool(0))
```
→ False

규칙:
#### 숫자
* 0 → False
* 나머지 숫자 → True
#### 문자열
```python
bool("hello")
```
→ True
```python
bool("")
```
→ False

빈 문자열만 False

#### Truthy / Falsy

Python에서는 값들을 자동으로 참/거짓처럼 판단합니다.
| 값          | 판단    |
| ---------- | ----- |
| `0`        | False |
| `""`       | False |
| 빈 리스트 `[]` | False |
| 나머지 대부분    | True  |

#### if문에서 자동 변환

```python
if 0:
    print(0)
elif "spam":
    print("spam")
```

출력:
```python
spam
```

왜냐하면:
* 0 → False
* "spam" → True

이기 때문입니다.
