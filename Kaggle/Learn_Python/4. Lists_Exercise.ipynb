#### 1번

L이라는 리스트가 들어옴
리스트의 두 번째 값(index 1) 을 반환
만약 리스트 길이가 2보다 작으면 None 반환

```python
def select_second(L):
    """주어진 리스트의 두 번째 요소를 반환한다.
    리스트에 두 번째 요소가 없으면 None을 반환한다.
    """
    if len(L) < 2:
        return None
    return L[1]
```

#### 3번

리스트 맨 앞 = 1등
리스트 맨 뒤 = 꼴등
둘의 위치를 교환(swap)

```python
def purple_shell(racers):
    """레이서 리스트에서
    1등(맨 앞)과 꼴등(맨 뒤)의 위치를 서로 바꿔라.
    """
    temp = racers[0]
    racers[0] = racers[-1]
    racers[-1] = temp
```

fashionably late 조건
1. 전체 사람의 절반보다 늦게 도착해야 함
2. 하지만 마지막 사람은 안 됨

```python
party_attendees = [
    'Adela', 'Fleda', 'Owen',
    'May', 'Mona', 'Gilbert', 'Ford'
]

def fashionably_late(arrivals, name):
    """파티 참석 순서 리스트와 이름이 주어졌을 때,
    그 사람이 'fashionably late'인지 반환하라.
    """
    
    order = arrivals.index(name)
    return order >= len(arrivals) / 2 and order != len(arrivals) - 1
```
