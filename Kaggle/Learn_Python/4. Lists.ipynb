#### Lists (리스트)

Python에서 리스트(List) 는 순서가 있는 값들의 모음입니다.

예시:
```python
primes = [2, 3, 5, 7]
```

리스트에는 숫자 말고도 다른 타입의 값들을 넣을 수 있습니다.

```python
planets = ['Mercury', 'Venus', 'Earth', 'Mars']
```

리스트 안에 또 다른 리스트를 넣을 수도 있습니다.

```python
hands = [
    ['J', 'Q', 'K'],
    ['2', '2', '2'],
    ['6', 'A', 'K'],
]
```

한 줄로도 작성 가능합니다.

```python
hands = [['J', 'Q', 'K'], ['2', '2', '2'], ['6', 'A', 'K']]
```

리스트에는 서로 다른 타입의 값들을 섞어서 넣을 수도 있습니다.

```python
my_favourite_things = [32, 'raindrops on roses', help]
```

#### Indexing (인덱싱)

대괄호 [] 를 사용해서 리스트의 특정 요소에 접근할 수 있습니다.

```python
planets[0]
```

결과:
```python
'Mercury'
```

Python은 0부터 시작하는 인덱스를 사용합니다.

* 첫 번째 요소 → index 0
* 두 번째 요소 → index 1

```python
planets[1]
```

결과:
```python
'Venus'
```

#### 음수 인덱스

뒤에서부터 접근할 수도 있습니다.

```python
planets[-1]
```

결과:
```python
'Neptune'
planets[-2]
```

결과:
```python
'Uranus'
```

* -1 → 마지막 요소
* -2 → 뒤에서 두 번째 요소

#### Slicing (슬라이싱)

리스트의 일부를 잘라낼 수 있습니다.

```python
planets[0:3]
```

결과:
```python
['Mercury', 'Venus', 'Earth']
```

의미:
* index 0부터 시작
* index 3 전까지 가져오기

즉:
```python
0 <= index < 3
```

#### 시작 인덱스 생략
```python
planets[:3]
```

결과:
```python
['Mercury', 'Venus', 'Earth']
```
시작 인덱스를 생략하면 자동으로 0이 됩니다.

#### 끝 인덱스 생략
```python
planets[3:]
```

결과:
```python
['Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune']
```

의미:
index 3부터 끝까지

#### 음수 슬라이싱
```python
planets[1:-1]
```

결과:
```python
['Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus']
```

의미:
첫 번째와 마지막 요소 제외

```python
planets[-3:]
```

결과:
```python
['Saturn', 'Uranus', 'Neptune']
```

의미:
마지막 3개 가져오기

#### Changing lists (리스트 수정)

리스트는 mutable(변경 가능) 합니다.
즉, 생성 후에도 값을 바꿀 수 있습니다.

```python
planets[3] = 'Malacandra'
```

결과:
```python
['Mercury',
 'Venus',
 'Earth',
 'Malacandra',
 'Jupiter',
 'Saturn',
 'Uranus',
 'Neptune']
```

#### 슬라이스를 이용한 수정

```python
planets[:3] = ['Mur', 'Vee', 'Ur']
```

결과:
```python
['Mur', 'Vee', 'Ur', 'Malacandra', ...]
```

다시 원래대로 복구:
```python
planets[:4] = ['Mercury', 'Venus', 'Earth', 'Mars']
```

#### List functions (리스트 함수)


#### len()

리스트 길이 반환

```python
len(planets)
```

결과:
```python
8
```

#### sorted()

정렬된 새 리스트 반환

```python
sorted(planets)
```

결과:
```python
['Earth', 'Jupiter', 'Mars', ...]
```

#### sum()

합계 계산

```python
primes = [2, 3, 5, 7]
sum(primes)
```

결과:
```python
17
```

#### max()

최댓값 반환

```python
max(primes)
```

결과:
```python
7
```

#### Objects (객체)

Python에서는 거의 모든 것이 객체(Object)입니다.

객체는:
* 데이터(속성)
* 함수(메서드)

를 함께 가지고 있습니다.

#### Attribute (속성)

예시:
```python
x = 12
x.imag
```

결과:
```python
0
```

imag는 복소수의 허수 부분입니다.

복소수 예시:
```python
c = 12 + 3j
c.imag
```

결과:
```python
3.0
```

#### Method (메서드)

객체 안에 포함된 함수를 메서드라고 합니다.

예시:
```python
x.bit_length()
```

결과:
```python
4
```

의미:
숫자를 이진수로 표현할 때 필요한 비트 수

#### help()

도움말 보기

```python
help(x.bit_length)
```

#### List methods (리스트 메서드)
#### append()

리스트 맨 뒤에 요소 추가

```python
planets.append('Pluto')
```
리스트 자체가 수정됩니다.

#### pop()

마지막 요소 제거 후 반환

```python
planets.pop()
```

결과:
```python
'Pluto'
```

#### Searching lists (리스트 검색)
#### index()

특정 값의 위치 찾기

```python
planets.index('Earth')
```

결과:
```python
2
```

없는 값을 찾으면 에러 발생:
```python
planets.index('Pluto')
```

결과:
```python
ValueError
```

#### in 연산자

값 존재 여부 확인
```python
"Earth" in planets
```

결과:
```python
True
```

```python
"Calbefraques" in planets
```
결과:
```python
False
```

#### Tuples (튜플)

튜플은 리스트와 거의 같지만 차이점이 2개 있습니다.

1. 괄호 사용
```python
t = (1, 2, 3)
```

또는:
```python
t = 1, 2, 3
```

2. 수정 불가능 (immutable)
```python
t[0] = 100
```

결과:
```python
TypeError
```

튜플은 생성 후 수정할 수 없습니다.
