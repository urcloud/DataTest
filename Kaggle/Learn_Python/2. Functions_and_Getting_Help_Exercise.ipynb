힌트(HINT): Python에는 round라는 내장 함수가 있습니다.

```python
def round_to_two_places(num):
    """주어진 숫자를 소수점 둘째 자리까지 반올림하여 반환합니다.
    
    >>> round_to_two_places(3.14159)
    3.14
    """
    return round(num, 2)

# 정답 확인
# q1.check()
```

음수일 경우 어떤 일이 일어날지 생각해보세요.
아래 셀에서 직접 여러 예제를 실행해보세요.

```python
help(round)

round(3.14, -2)
```

* ndigits=-1 이면 가장 가까운 10의 자리로 반올림
* ndigits=-2 이면 가장 가까운 100의 자리로 반올림
하는 식으로 동작합니다.

예를 들어 아주 큰 숫자를 다룬다고 가정해봅시다.

* 핀란드의 면적은 338,424 km²
* 그린란드의 면적은 2,166,086 km² 입니다.

하지만 실제로는:
* 338,424인지
* 338,425인지
* 338,177인지

같은 세부 숫자는 크게 중요하지 않을 수 있습니다.
오히려 너무 많은 자릿수는 정보를 보기 어렵게 만들 수 있습니다.
이럴 때 round() 함수에 ndigits=-3 을 사용하면 뒤의 숫자를 잘라낼 수 있습니다.

예시:
* 핀란드의 면적은 338,000 km²
* 그린란드의 면적은 2,166,000 km²
처럼 표현할 수 있습니다.

### 3번

이전 문제에서 사탕을 나눠 가지던 Alice, Bob, Carol은 사탕을 공평하게 나누려고 했습니다.
남는 사탕은 우정을 위해 부숴버리기로 했습니다.

예를 들어 사탕이 총 91개라면:
* 각자 30개씩 받고
* 1개는 남아서 부숴야 합니다.

아래 함수는 남는 사탕 개수를 계산합니다.

이 함수를 수정해서,
사탕을 나눌 친구 수를 두 번째 인자로 선택적으로 받을 수 있게 만드세요.

만약 두 번째 인자를 주지 않으면 기존처럼 친구 수를 3명으로 가정해야 합니다.

또한 docstring도 새로운 동작에 맞게 수정하세요.

```python
def to_smash(total_candies, n=3):
    """주어진 사탕을 3명의 친구에게 똑같이 나누고 남아서
    부숴야 하는 사탕 개수를 반환합니다.
    
    >>> to_smash(91)
    1
    """

    return total_candies % n

# 정답 확인
# q3.check()
```
